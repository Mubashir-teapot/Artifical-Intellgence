```python
import requests

OLLAMA_URL = "http://localhost:11434"
MODEL = "qwen2.5:0.5b"


def local_ai_agent(user_input):
    prompt = f"""
You are a university Artificial Intelligence lab assistant.

Answer clearly and simply.

User Question:
{user_input}

Answer:
"""

    response = requests.post(
        f"{OLLAMA_URL}/api/generate",
        json={
            "model": MODEL,
            "prompt": prompt,
            "stream": False
        },
        timeout=120
    )

    response.raise_for_status()

    return response.json()["response"]


print("Local AI Agent Started")
print("Type 'exit' to stop\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Agent stopped.")
        break

    try:
        answer = local_ai_agent(user)

        print("\nAI:")
        print(answer)
        print()

    except Exception as e:
        print("Error:", e)
        print("Make sure Ollama is running with: ollama serve")
```
