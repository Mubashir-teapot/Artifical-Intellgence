```python
# ============================================
# INTELLIGENT TRAVEL AGENT
# Agent Type: Utility-Based Agent
# ============================================

# Sample Flight Database
flights = [
    {
        "id": "PK101",
        "from": "Karachi",
        "to": "Lahore",
        "price": 18000,
        "duration": 1.8
    },
    {
        "id": "PK102",
        "from": "Karachi",
        "to": "Lahore",
        "price": 15000,
        "duration": 2.0
    },
    {
        "id": "PK103",
        "from": "Karachi",
        "to": "Lahore",
        "price": 21000,
        "duration": 1.6
    },
    {
        "id": "PK201",
        "from": "Karachi",
        "to": "Islamabad",
        "price": 20000,
        "duration": 2.0
    },
    {
        "id": "PK202",
        "from": "Karachi",
        "to": "Islamabad",
        "price": 17500,
        "duration": 2.3
    },
    {
        "id": "PK301",
        "from": "Lahore",
        "to": "Islamabad",
        "price": 12000,
        "duration": 1.0
    },
    {
        "id": "PK302",
        "from": "Lahore",
        "to": "Islamabad",
        "price": 10000,
        "duration": 1.2
    }
]


class TravelAgent:

    def __init__(self, flights):
        self.flights = flights

    # -----------------------------
    # SENSOR / PERCEPTION
    # -----------------------------
    def get_user_input(self):

        source = input("Enter departure city: ").strip().title()
        destination = input("Enter destination city: ").strip().title()

        while True:
            try:
                budget = float(input("Enter maximum budget (PKR): "))

                if budget <= 0:
                    print("Budget must be greater than 0.")
                    continue

                break

            except ValueError:
                print("Please enter a valid number.")

        preference = input(
            "Enter preference (cheapest/fastest): "
        ).strip().lower()

        if preference not in ["cheapest", "fastest"]:
            print("Invalid preference.")
            print("Cheapest selected by default.")
            preference = "cheapest"

        return source, destination, budget, preference

    # -----------------------------
    # SEARCH AVAILABLE FLIGHTS
    # -----------------------------
    def search_flights(self, source, destination, budget):

        available_flights = []

        for flight in self.flights:

            if (
                flight["from"].lower() == source.lower()
                and flight["to"].lower() == destination.lower()
                and flight["price"] <= budget
            ):
                available_flights.append(flight.copy())

        return available_flights

    # -----------------------------
    # DECISION MAKING
    # UTILITY-BASED AGENT
    # -----------------------------
    def select_best_flight(self, available_flights, preference):

        if len(available_flights) == 0:
            return None

        prices = [flight["price"] for flight in available_flights]
        durations = [flight["duration"] for flight in available_flights]

        minimum_price = min(prices)
        maximum_price = max(prices)

        minimum_duration = min(durations)
        maximum_duration = max(durations)

        for flight in available_flights:

            # Price utility
            if maximum_price == minimum_price:
                price_score = 1
            else:
                price_score = (
                    maximum_price - flight["price"]
                ) / (
                    maximum_price - minimum_price
                )

            # Duration utility
            if maximum_duration == minimum_duration:
                duration_score = 1
            else:
                duration_score = (
                    maximum_duration - flight["duration"]
                ) / (
                    maximum_duration - minimum_duration
                )

            # Utility function
            if preference == "cheapest":

                flight["utility"] = (
                    0.7 * price_score
                    + 0.3 * duration_score
                )

            else:

                flight["utility"] = (
                    0.3 * price_score
                    + 0.7 * duration_score
                )

        best_flight = max(
            available_flights,
            key=lambda flight: flight["utility"]
        )

        return best_flight

    # -----------------------------
    # ACTUATOR / ACTION
    # -----------------------------
    def display_result(self, flight):

        if flight is None:

            print("\nNo suitable flight found.")
            return

        print("\n==============================")
        print("     RECOMMENDED FLIGHT")
        print("==============================")

        print("Flight ID :", flight["id"])
        print("From      :", flight["from"])
        print("To        :", flight["to"])
        print("Price     : PKR", flight["price"])
        print("Duration  :", flight["duration"], "hours")
        print("Utility   :", round(flight["utility"], 2))

        booking = input(
            "\nDo you want to book this flight? (yes/no): "
        ).strip().lower()

        if booking in ["yes", "y"]:

            print("\nFlight booked successfully!")
            print("Booking ID:", flight["id"])

        else:

            print("\nBooking cancelled.")

    # -----------------------------
    # RUN AGENT
    # -----------------------------
    def run(self):

        print("==============================")
        print("   INTELLIGENT TRAVEL AGENT")
        print("==============================")

        # Sensor input
        source, destination, budget, preference = self.get_user_input()

        # Search environment
        available_flights = self.search_flights(
            source,
            destination,
            budget
        )

        # Decision
        best_flight = self.select_best_flight(
            available_flights,
            preference
        )

        # Action
        self.display_result(best_flight)


# ============================================
# START AGENT
# ============================================

agent = TravelAgent(flights)

agent.run()
```
